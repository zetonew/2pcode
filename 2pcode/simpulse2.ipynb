```python
import numpy as np  
import pandas as pd  
from PIL import Image  
import os
  
# 调用函数  
image_path = '/Data_Processing/ZL/2P/M2/C2/M2-9-2.oir.C2.tif'  #文件读取路径
csvname = 'M2-9-2.oir.C2.tif'#文件名
csv_file_path = '/Data_Processing/ZL/2P/M2/C2_RES/' #保存路径


# 打开TIFF文件  

img = Image.open(image_path)
# 初始化一个列表来存储每帧的亮度  
#print(({image_path})) 

num_frames = img.n_frames-1  
print(({num_frames})) 
# 初始化一个数组来存储每帧的亮度  
brightness_values = np.zeros(num_frames)  
    
# 遍历每一帧  
for i in range(num_frames):  
    img.seek(i)  # 移动到指定帧  
    frame = np.array(img)  # 将帧转换为NumPy数组  
    # 计算亮度  
    brightness_values[i] = np.mean(frame)  
    #brightness_values.append(brightness)  # 将亮度值添加到列表中  


# 将亮度值列表转换为pandas DataFrame  
df = pd.DataFrame(brightness_values, columns=['Brightness'])  

# 存储为CSV文件  
#csv_file_path = 'brightness_values.csv'  
df.to_csv(f'{csv_file_path}{csvname}.csv', index=False)  # index=False表示不将行索引写入CSV文件  
print(f'({image_path}) save to ({csv_file_path})')
```

批处理

```python
import numpy as np  
import pandas as pd  
from PIL import Image  
import os
  
# 调用函数  
filepath = '/Data_Processing/ZL/2P/M3/C2'  #读取文件所在文件夹
savepath = '/Data_Processing/ZL/2P/M3/C2_RES/' #保存路径

def calculate_brightness(image_path,csv_file_path,csvname):  
    # 打开TIFF文件  
    
    img = Image.open(image_path)
    # 初始化一个列表来存储每帧的亮度  
    #print(({image_path})) 
    
    num_frames = img.n_frames-1  
    print(({num_frames})) 
    # 初始化一个数组来存储每帧的亮度  
    brightness_values = np.zeros(num_frames)  
        
    # 遍历每一帧  
    for i in range(num_frames):  
        img.seek(i)  # 移动到指定帧  
        frame = np.array(img)  # 将帧转换为NumPy数组  
        # 计算亮度  
        brightness_values[i] = np.mean(frame)  
        #brightness_values.append(brightness)  # 将亮度值添加到列表中  

  
    # 将亮度值列表转换为pandas DataFrame  
    df = pd.DataFrame(brightness_values, columns=['Brightness'])  
  
    # 存储为CSV文件  
    #csv_file_path = 'brightness_values.csv'  
    df.to_csv(f'{csv_file_path}{csvname}.csv', index=False)  # index=False表示不将行索引写入CSV文件  
    print(f'({image_path}) save to ({csv_file_path})')
```

```python
def process_files_in_directory(filepath,savepath):  
    """  
    遍历指定目录下的所有.tif文件。  
    """  
    # 遍历目录  
    for filename in os.listdir(filepath):  
        # 检查文件是否以.txt结尾  
        if filename.endswith(".tif"):  
            # 构建完整的文件路径
            csvname =  filename
            filename = os.path.join(filepath, filename)  
            
            if __name__ == "__main__":
                calculate_brightness(filename,savepath,csvname)
           
  
# 使用函数  

process_files_in_directory(filepath,savepath)
```
